I viewed the columns to check that feature names were properly adjusted.

```python
# Move separate columns ('male_archery', 'male_badminton') under a single 'Event' column, with their corresponding medal type in a new 'Medal' column.
    # .melt reshapes the dataframe from wide to long by designating an 'Event' and 'Medal' column
df_cleaned = (df
            .melt(id_vars = ['Medalist Name'],
                    value_vars = ['Male Archery', 'Female Archery', 'Male Athletics',
                                  'Female Athletics', 'Male Badminton', 'Female Badminton',
                                  'Male Baseball', 'Male Basketball', 'Female Basketball', 'Male Boxing',
                                  'Male Canoeing and Kayaking', 'Female Canoeing and Kayaking',
                                  'Male Road Bicycle Racing', 'Female Road Bicycle Racing',
                                  'Male Track Cycling', 'Female Track Cycling', 'Male Mountain Biking',
                                  'Female Mountain Biking', 'Male BMX', 'Female BMX', 'Male Diving',
                                  'Female Diving', 'Female Equestrian Sport', 'Male Equestrian Sport',
                                  'Male Fencing', 'Female Fencing', 'Male Field Hockey',
                                  'Female Field Hockey', 'Male Association Football',
                                  'Female Association Football', 'Male Artistic Gymnastics',
                                  'Female Artistic Gymnastics', 'Female Rhythmic Gymnastics',
                                  'Male Trampoline Gymnastics', 'Female Trampoline Gymnastics',
                                  'Male Handball', 'Female Handball', 'Male Judo', 'Female Judo',
                                  'Male Modern Pentathlon', 'Female Modern Pentathlon', 'Male Rowing',
                                  'Female Rowing', 'Male Sailing', 'Female Sailing',
                                  'Male Shooting Sport', 'Female Shooting Sport', 'Female Softball',
                                  'Male Swimming', 'Female Swimming', 'Female Synchronized Swimming',
                                  'Male Table Tennis', 'Female Table Tennis', 'Male Taekwondo',
                                  'Female Taekwondo', 'Male Tennis', 'Female Tennis', 'Male Triathlon',
                                  'Female Triathlon', 'Male Beach Volleyball', 'Female Beach Volleyball',
                                  'Male Volleyball', 'Female Volleyball', 'Male Water Polo',
                                  'Female Water Polo', 'Male Weightlifting', 'Female Weightlifting',
                                  'Male Freestyle Wrestling', 'Female Freestyle Wrestling', 'Male Greco-Roman Wrestling'],
                    var_name = 'Event',
                    value_name = 'Medal'
                    )
           .dropna(subset = ['Medal']) # Remove rows where medal is missing (does not include extra entries for events athletes did not compete in)
)

# Split sport and sex into separate features to represent each variable in its own column
df_cleaned[['Sex', 'Sport']] = df_cleaned['Event'].str.split(n = 1, expand = True) # Splits each string at spaces: since sex is the first word before the space, splitting at the first space effectively separates sex and sport designations, and turning the results into separate columns
df_cleaned = df_cleaned.drop(columns = 'Event') # Drop 'Event' to eliminate redundancy

# Standardize capitalization
df_cleaned['Medal'] = df_cleaned['Medal'].str.capitalize() 

# Define medal type as a categorical variable with designated order
df_cleaned['Medal'] = pd.Categorical(
    df_cleaned['Medal'],
    categories =['Gold', 'Silver', 'Bronze'],
    ordered = True)

# Sort entries by sport, then sex, then medal
df_cleaned = df_cleaned.sort_values(['Sport', 'Sex', 'Medal'])

# Visualize cleaned layout
df_cleaned.head()
```

## Why the Tidy Data Process was Important for this Dataset

The first step to transform the data into a tidy format was to reshape it from its original wide format to a long format. The .melt function in the Pandas library accomplishes this by combining multiple, descriptive features under a single variable column and placing their corresponding values in a new, designated column, keeping associated observations in the same row. This reorganizes the data so that each variable has its own column, and each observation has its own row. In the Olympics dataset, each sport-sex combination was represented by a unique column, which produced an extremely wide dataset. Medal type observations were spread out across many of the sport-sex columns, with missing values under each sport-sex event that the athlete did not compete in. Using the melt function was essential to tidy the data because it collapsed sport-sex columns under a single 'Event' column, while creating a unique column to represent the 'Medal' variable. This transformed the data from a wide to long format, reducing missing values and ensuring variables were appropriately represented by combining events under a single, inclusive column and designating the medal type variable as its own column.

After melting the dataset, it still did not completely meet tidy data standards because the sport-sex combination represented the variables of 'Sex' and 'Sport' together, rather than designating a unique column for each variable. The .split function was used to split the 'Sex' designation from the 'Sport' designation in the 'Event' observations, forming new columns to represent each variable and better uphold this principle of tidy data. I was able to split the 'Event' observation strings at the first space since each event began with either a 'Male' or 'Female' designation. This step allowed me to create separate columns for sport and gender designations, upholding the tidy data principle that each variable should have its own column.

After these steps, the data was further organized through the process of designating medal ranks and further sorting the data by sport, sex, and medal.

Overall, this process produced a more clean and structured dataset, ready for further analysis and visualization.

## Objective 2
### Visualization:
- Include at least two visualizations that reflects the newly cleaned data.

```python
# Import libraries for visualization
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Create a bar chart showing medals by sport, ordered from greatest to least number of medals
# Count medals per sport, and sort from most medals to least values
medals_per_sport = df_cleaned.groupby('Sport')['Medal'].count().sort_values(ascending = False)
df_medals_per_sport = pd.DataFrame(medals_per_sport)

plt.figure(figsize = (12,6)) # Enlargen figure
sns.barplot(df_medals_per_sport, x = 'Sport', y = 'Medal') # Designate variables to use for x and y axes
plt.xticks(rotation = 90) # Make X labels vertical (to improve readability)
plt.title("Total Medals by Sport")
plt.ylabel("Number of Medals")
plt.xlabel("Sport")
plt.show()
```

```python
# Create a bar plot that counts and displays number of medals of each type by gender
plt.figure(figsize = (10,6)) # Enlargen figure
sns.countplot(
    data = df_cleaned,
    x = 'Sex', # Set independent variable as sex, countplot function counts instances of each sex
    hue = 'Medal', # Split counts for each sex by type of medal earned
    palette = {'Gold': 'gold', 'Silver': 'lightgray', 'Bronze': 'chocolate'} # Color code medals
)
plt.title("Medal Counts by Sex")
plt.ylabel("Number of Medals")
plt.xlabel("Sex")
plt.legend(title = "Medal Type")
plt.show()
```

## Visualizations

Transforming the dataset into a tidy format facilitated the process of creating clear and organized visualizations. The Seaborn and MatPlotLib libraries are extremely effective tools for working with tidy data since their functions allow extensive, multivariate visualization of tidy data that has clearly defined variables.

The first visualization displays the total number of medals by sport. The cleaned dataset allowed medals to be grouped by sport and plotted using a bar chart. Since each observation represents a single medal and the 'Sport' variable is in its own column, grouping and counting medals was facilitated by the tidy format of the data. The second visualization displays medal counts by sex, with each medal type represented by a unique color. This chart demonstrates how tidy data labels information clearly so that it can be accurately visualized in a multivariate representation. Together, these visualizations show how organizing data into a tidy format allows for more intuitive aggregation, counting, and plotting, ultimately making data patterns more accessible.

## Objective 3
### Pivot-Table:
- Run at least one aggregation function on the tidy data.

```python
# Create a pivot table to display a count of medals earned by each sex in each sport
pivot_medals = df_cleaned.pivot_table(
    index = 'Sport', # Designate sport as index
    columns = 'Sex', # Split sex into 'Female' and 'Male' columns
    values = 'Medal', # Designate 'Medal' instances as the values to aggregate
    aggfunc = 'count', # Task is to count medals under each column
    fill_value = 0 # Fill missing values with 0 (i.e. no females won baseball medals, so puts zero in place of missing value)
)

# Add total medals column
pivot_medals['Total Medals'] = pivot_medals.sum(axis = 1) # calculate sum across rows (axis = 1 directs sum function to add Male + Female values), and display under 'Total medals' feature

# Sort by descending total medals
pivot_medals = pivot_medals.sort_values('Total Medals', ascending = False)

# View table
pivot_medals
```

## Pivot Table and Aggregation

Similarly to visualization, tidy formats help organize the data to make aggregation and summarization more straightforward. By giving each column its own variable, the data can be grouped and organized to present information and allow different patterns to emerge.

Here, I created a pivot table to display the number of medals earned by each sex within each sport. This visualization offers a bridge between the two visualizations I created earlier. Instead of displaying overall medal counts by sport or medal counts by gender, this pivot table demonstrates medals by gender for each sport at the same time. Tidy data format makes this pivot table simple to create since sport and sex are in their own columns and can be more readily counted as a result. The pivot table provides an organized summary of the dataset and allows viewers to identify some patterns almost immediately even without visualization tools, such as gender-related discrepancies in medal earnings. This demonstrates how tidy data can facilitate summarization of data to readily provide viewers with general data insights.
